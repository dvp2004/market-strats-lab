```python
from pathlib import Path
import pandas as pd

ROOT = Path("..").resolve()
DATA = {
    "candidate_summary": ROOT
    / "reports/strategy_factory/regime_stress/phase21a_candidate_regime_summary.csv",
    "robustness_scores": ROOT
    / "reports/strategy_factory/regime_stress/phase21a_regime_robustness_scores.csv",
    "score_components": ROOT
    / "reports/strategy_factory/regime_stress/phase21a_regime_robustness_score_components.csv",
    "regime_metrics": ROOT / "reports/strategy_factory/regime_stress/phase21a_regime_metrics.csv",
    "shortlist": ROOT
    / "reports/strategy_factory/regime_reconciliation/phase21b_paper_shortlist_recommendation.csv",
    "targets": ROOT
    / "reports/paper_trading/regime_informed_tracking/regime_informed_paper_targets.csv",
    "tear_sheet": ROOT
    / "reports/paper_trading/regime_informed_tracking/regime_informed_daily_tracking_tear_sheet.csv",
    "discipline": ROOT
    / "reports/paper_trading/regime_informed_tracking/regime_informed_session_discipline_summary.csv",
    "ledger": ROOT
    / "reports/paper_trading/regime_informed_tracking/regime_informed_manual_session_ledger.csv",
    "session_ingestion_status": ROOT
    / "reports/paper_trading/dashboard/regime_informed_session_ingestion_status.csv",
    "daily_runtime": ROOT / "reports/paper_trading/dashboard/daily_paper_runtime_status.csv",
}


def read_csv(path):
    if not path.exists():
        return pd.DataFrame({"missing_file": [str(path)]})
    return pd.read_csv(path)


frames = {name: read_csv(path) for name, path in DATA.items()}
pd.DataFrame(
    {
        "source": list(DATA),
        "path": [str(path) for path in DATA.values()],
        "exists": [path.exists() for path in DATA.values()],
    }
)
```
